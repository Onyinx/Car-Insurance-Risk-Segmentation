```python
import pandas as pd
from scipy import stats
```

```python
_path = "/content/sample_data/Car_Insurance_Source.xlsx"
df = pd.read_excel(_path)
print(f"Loaded: {len(df):,} policies\n")
```

```python
#Do the average claim amounts for each group
#in this variable actually differ, or could the difference just be random
#chance?

print("=== Testing each variable against claim amount ===")
print("")

# --- Coverage Zone ---
groups = [group["claim_amt"] for name, group in df.groupby("coverage_zone")]
f_stat, p_value = stats.f_oneway(*groups)
print("Q1. Coverage Zone:")
print("  F-statistic =", round(f_stat, 2))
print("  p-value =", round(p_value, 4))
print("  Significant?", "YES" if p_value < 0.05 else "NO")
print("")

# --- Marital Status ---
groups = [group["claim_amt"] for name, group in df.groupby("marital_status")]
f_stat, p_value = stats.f_oneway(*groups)
print("Q1. Marital Status:")
print("  F-statistic =", round(f_stat, 2))
print("  p-value =", round(p_value, 4))
print("  Significant?", "YES" if p_value < 0.05 else "NO")
print("")

# --- Education ---
groups = [group["claim_amt"] for name, group in df.groupby("education")]
f_stat, p_value = stats.f_oneway(*groups)
print("Q1. Education:")
print("  F-statistic =", round(f_stat, 2))
print("  p-value =", round(p_value, 4))
print("  Significant?", "YES" if p_value < 0.05 else "NO")
print("")

# --- Car Color ---
groups = [group["claim_amt"] for name, group in df.groupby("car_color")]
f_stat, p_value = stats.f_oneway(*groups)
print("Q4. Car Color:")
print("  F-statistic =", round(f_stat, 2))
print("  p-value =", round(p_value, 4))
print("  Significant?", "YES" if p_value < 0.05 else "NO")
print("")

# --- Gender ---
groups = [group["claim_amt"] for name, group in df.groupby("gender")]
f_stat, p_value = stats.f_oneway(*groups)
print("Q1. Gender:")
print("  F-statistic =", round(f_stat, 2))
print("  p-value =", round(p_value, 4))
print("  Significant?", "YES" if p_value < 0.05 else "NO")
print("")

# --- Car Use ---
groups = [group["claim_amt"] for name, group in df.groupby("car_use")]
f_stat, p_value = stats.f_oneway(*groups)
print("Q2. Car Use:")
print("  F-statistic =", round(f_stat, 2))
print("  p-value =", round(p_value, 4))
print("  Significant?", "YES" if p_value < 0.05 else "NO")
print("")
```

```python
# Step 3: Correlation tests
# A correlation close to 0 means "no relationship." A p-value under 0.05
# means the relationship (even a weak one) is real, not random chance.

print("=== Testing income and vehicle year against claims ===")
print("")

correlation, p_value = stats.pearsonr(df["household_income"], df["claim_amt"])
print("Q5. Household Income vs Claim Amount:")
print("  correlation =", round(correlation, 4))
print("  p-value =", round(p_value, 4))
print("  Significant?", "YES" if p_value < 0.05 else "NO")
print("")

correlation, p_value = stats.pearsonr(df["car_year"], df["claim_amt"])
print("Q5 (bonus). Vehicle Year vs Claim Amount:")
print("  correlation =", round(correlation, 4))
print("  p-value =", round(p_value, 4))
print("  Significant?", "YES" if p_value < 0.05 else "NO")
print("")
```

```python
# Step 4: Car make — but only for makes with enough policies to trust

# Some car makes only have a handful of policies. Averaging 3 policies and
# calling it a "finding" is misleading, so we only look at makes with at
# least 30 policies.

print("=== Q3. Car make (only makes with 30+ policies) ===")
print("")

make_counts = df.groupby("car_make").size()
reliable_makes = make_counts[make_counts >= 30].index
reliable_df = df[df["car_make"].isin(reliable_makes)]
make_averages = reliable_df.groupby("car_make")["claim_amt"].mean().round(2)
print(make_averages.sort_values(ascending=False))
print("")
```

Conclusion

Every test above comes back not significant (p > 0.05)  coverage zone, marital status, education, car color, gender, car use, household income, and vehicle year all fail to show a real relationship with claim amount. This isn't a gap in the analysis; it's a genuine finding. The honest recommendation is to stop pricing based on these variables and instead invest in collecting real risk data driving record, annual mileage, and prior claims history  since none of the variables currently available actually separate high-risk policyholders from low-risk ones
